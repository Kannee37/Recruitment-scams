## Mã hóa trường dữ liệu

```python
# Cột nhãn, 1 là Lừa đảo, 0 là thật
df_temp['Nhãn'] = df_temp['Nhãn'].map(lambda x: 0 if x == 'Thật' else 1)
df_temp['Nhãn'].value_counts()
```

```python
# Cột tài khoản
df_temp['Tài khoản'] = df_temp['Tài khoản'].map(lambda x: 0 if x == 'Không' else 1)
df_temp['Tài khoản'].value_counts()
```

```python
# Cột mô tả công việc
df_temp['Mô tả công việc'] = df_temp['Mô tả công việc'].map(lambda x: 0 if x == 'Không' else 1)
df_temp['Mô tả công việc'].value_counts()
```

```python
df_temp['Nhóm ngành nghề'].value_counts().sort_index()
```

```python
# Dùng Label Encoding cho cột Nhóm ngành nghề bởi mỗi dòng chỉ thuộc về 1 nhóm ngành
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_temp['Nhóm ngành nghề'] = le.fit_transform(df_temp['Nhóm ngành nghề'])

df_temp['Nhóm ngành nghề'].value_counts().sort_index()
```

```python
# Dùng one-hot cho các cột hình thức nhận việc, hình thức làm
df_temp['Hình thức làm'] = df_temp['Hình thức làm'].str.lower().str.replace(' ','').str.strip()
df_temp['Hình thức nhận việc'] = df_temp['Hình thức nhận việc'].str.lower().str.replace(' ','').str.strip()

df_temp['Hình thức làm'] = df_temp['Hình thức làm'].str.split(',')
df_temp['Hình thức nhận việc'] = df_temp['Hình thức nhận việc'].str.split(',')
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

labels_column = df_temp['Hình thức nhận việc']

mlb = MultiLabelBinarizer()
encoded = mlb.fit_transform(labels_column)
df_encoded = pd.DataFrame(encoded, columns=mlb.classes_, index=df_temp.index)
df_encoded.columns = ['Hình thức nhận việc - ' + col for col in df_encoded.columns]
df_encoded
```

```python
df_temp_encoded = pd.concat([df_temp.iloc[:,:-2], df_encoded], axis=1)
df_temp_encoded.head()
```

```python
labels_column = df_temp['Hình thức làm']

mlb = MultiLabelBinarizer()
encoded = mlb.fit_transform(labels_column)
df_encoded = pd.DataFrame(encoded, columns=mlb.classes_, index=df_temp.index)
df_encoded.columns = ['Hình thức làm - ' + col for col in df_encoded.columns]
df_encoded
```

```python
df_temp_encoded = pd.concat([df_temp_encoded, df_encoded], axis=1)
df_temp_encoded.head()
```

```python
df_temp = pd.concat([df_temp_encoded, df_temp.iloc[:,-2:]], axis=1)
```

```python
df_temp.drop(['Hình thức làm', 'Hình thức nhận việc'], axis=1, inplace=True)
```
